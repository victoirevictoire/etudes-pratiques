# Système d'Analyse de Documents Electroniques
Ce Notebook intègre la détection de zones (YOLO), l'extraction de nomenclature (EasyOCR) et la localisation de composants.

```python
import os, cv2, sys
import numpy as np
import ipywidgets as widgets
from IPython.display import display, clear_output
from ultralytics import YOLO
import easyocr
import matplotlib.pyplot as plt
from pdf2image import convert_from_bytes

# Chemins vers tes modèles (à adapter si besoin)
PATH_ZONES = "best.pt"
PATH_REFS  = "bestr.pt"

print("Initialisation des modèles...")
model_zones = YOLO(PATH_ZONES)
model_refs  = YOLO(PATH_REFS)
reader = easyocr.Reader(['en', 'fr'], gpu=True)
clear_output()
print("Moteurs prêts.")

def extraire_zones(img_rgb):
    results = model_zones.predict(source=img_rgb, conf=0.212, verbose=False)
    zones = {"Table": None, "Diagram": None, "Board": None}
    margin = 10
    for box in results[0].boxes:
        label = model_zones.names[int(box.cls[0])]
        if label in zones:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            y1_m, y2_m = max(0, y1-margin), min(img_rgb.shape[0], y2+margin)
            x1_m, x2_m = max(0, x1-margin), min(img_rgb.shape[1], x2+margin)
            zones[label] = img_rgb[y1_m:y2_m, x1_m:x2_m]
    return zones

uploader = widgets.FileUpload(accept='.pdf, .png, .jpg', multiple=False)
out = widgets.Output()

def on_upload(change):
    with out:
        clear_output()
        if not uploader.value: return
        file_info = list(uploader.value.values())[0]
        content = file_info['content']
        
        if file_info['metadata']['name'].lower().endswith('.pdf'):
            images = convert_from_bytes(content, dpi=300)
            img_rgb = np.array(images[0])
        else:
            nparr = np.frombuffer(content, np.uint8)
            img_rgb = cv2.cvtColor(cv2.imdecode(nparr, cv2.IMREAD_COLOR), cv2.COLOR_BGR2RGB)

        print("Analyse en cours...")
        zones_detectees = extraire_zones(img_rgb)
        
        found = [k for k, v in zones_detectees.items() if v is not None]
        if not found:
            print("Aucune zone détectée.")
            return

        fig, axes = plt.subplots(1, len(found), figsize=(15, 7))
        if len(found) == 1: axes = [axes]
        for ax, z_name in zip(axes, found):
            ax.imshow(zones_detectees[z_name])
            ax.set_title(z_name)
            ax.axis('off')
        plt.show()

uploader.observe(on_upload, names='value')
display(widgets.VBox([widgets.HTML("<b>Upload de document</b>"), uploader, out]))
```
