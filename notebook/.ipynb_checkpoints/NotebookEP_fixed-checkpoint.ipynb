# Notebook EP - Version corrigée

```
# Imports principaux
import os
from merge import process_and_visualize
from extract_easyocr import extract, draw_results
import cv2
```

## Running the notebook with Voila

```
# === PARAMÈTRES ===
PDF_PATH = "input.pdf"
MODEL_PATH = "best.pt"

# === ÉTAPE 1 : Détection et extraction des zones ===
process_and_visualize(PDF_PATH, MODEL_PATH)

# === ÉTAPE 2 : OCR sur la carte extraite ===
if os.path.exists("temp_board.png"):
    img, detections = extract("temp_board.png", MODEL_PATH)

    annotated = draw_results(img, detections)
    cv2.imwrite("final_annotated.png", annotated)

    print("Résultat sauvegardé : final_annotated.png")
else:
    print("Erreur : temp_board.png introuvable")
```
